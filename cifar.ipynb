```python
from keras.datasets import cifar10
(x_train, y_train),(x_test, y_test)=cifar10.load_data()
x_train.shape
x_test.shape
x_train = x_train.reshape(50000, 32*32*3)
x_train= x_train.astype('float32')/255
x_test = x_test.reshape(10000, 32*32*3)
x_test= x_test.astype('float32')/255
x_test = x_test.reshape(10000, 32*32*3)
x_test= x_test.astype('float32')/255
from keras.utils import to_categorical
y_train = to_categorical(y_train, num_classes=10)
y_test = to_categorical(y_test, num_classes=10)
```

```python
from keras.models import Sequential
from keras.layers import Dense
model = Sequential()
model.add(Dense(512, input_shape=(32*32*3,), activation='relu'))
model.add(Dense(112, input_shape=(32*32*3,), activation='relu'))
model.add(Dense(10, input_shape=(32*32*3,), activation='softmax'))
model.summary()
```

```python
model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
history = model.fit(x_train, y_train, batch_size=64, epochs=50, validation_data=(x_test, y_test))
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from keras.models import load_model

class_names = ['airplane', 'automobile', 'bird', 'cat', 'deer', 'dog', 'frog', 'horse', 'plane', 'truck']

def preprocess_image(image_path):
    img = cv2.imread(image_path)

    if img is None:
        raise FileNotFoundError(f"Không tìm thấy ảnh tại đường dẫn: {image_path}")

    img = cv2.resize(img, (32, 32))
    img = img.astype("float32") / 255.0
    img = img.reshape(1, 32*32*3)
    return img

def predict_animal(image_path):
    processed_img = preprocess_image(image_path)
    prediction = model.predict(processed_img)
    predicted_class = np.argmax(prediction)
    return predicted_class

image_paths = [
    "/content/bird.webp",
    "/content/cat.jpg",
    "/content/deer.jpg",
    "/content/dog.png",
    "/content/frog.jpg",
    "/content/hourse.jpg",
    "/content/mobile.jpg",
    "/content/plane.jpg",
    "/content/truck.webp",
    "/content/tải xuống (1).jpg"
]

predicted_results = {}
for image_path in image_paths:
    predicted_results[image_path] = predict_animal(image_path)

for image_path, predicted_class in predicted_results.items():
    predicted_label = class_names[predicted_class]
    print(f"Image: {image_path}, Predicted Class: {predicted_label}")

    img = cv2.imread(image_path)
    plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
    plt.title(f"Predicted Animal Class: {predicted_label}")
    plt.axis("off")
    plt.show()
```
