```python
from keras.datasets import mnist
(x_train, y_train),(x_test, y_test)=mnist.load_data()
x_train.shape
x_test.shape
```

```python
for i in range(25):
  plt.subplot(5,5,i+1)
  plt.tight_layout()
  plt.imshow(x_train[5+i], cmap='gray', interpolation='none')
  plt.title("Số: {}".format(y_train[5+i]))
  plt.xticks([])
  plt.yticks([])
```

```python
import matplotlib.pyplot as plt
plt.imshow(x_train[8])
plt.show()
print('Số này là:', y_train[100])
```

```python
x_train = x_train.reshape(60000, 28*28)
x_train= x_train.astype('float32')/255
x_test = x_test.reshape(10000, 28*28)
x_test= x_test.astype('float32')/255
```

```python
from keras.utils import to_categorical
y_train = to_categorical(y_train)
y_test = to_categorical(y_test)
```

```python
from keras.models import Sequential
from keras.layers import Dense
model = Sequential()
model.add(Dense(512, input_shape=(784,), activation='relu'))
model.add(Dense(112, activation='relu'))
model.add(Dense(10, activation='softmax'))
model.summary()
```

```python
model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
```

```python
history = model.fit(x_train, y_train, batch_size=64, epochs=100, validation_data=(x_test, y_test))
model.save("/content/1.h5")
```

```python
import cv2
import cv2
import numpy as np
import matplotlib.pyplot as plt
from keras.models import load_model

model = load_model("/content/sample_data/1.h5")

def preprocess_image(image_path):
    """Tiền xử lý ảnh để phù hợp với mô hình MNIST."""
    img = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)
    if img is None:
        raise FileNotFoundError(f"Không tìm thấy ảnh tại đường dẫn: {image_path}")
    img = cv2.resize(img, (28, 28))
    img = cv2.bitwise_not(img)
    img = img.astype("float32") / 255.0
    img = img.reshape(28 * 28)
    return img

def predict_digit(image_path):
    """Dự đoán chữ số từ ảnh."""
    processed_img = preprocess_image(image_path)
    processed_img = np.expand_dims(processed_img, axis=0)
    prediction = model.predict(processed_img)
    predicted_digit = np.argmax(prediction)

    img = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)
    plt.imshow(img, cmap='gray')
    plt.title(f"Predicted Digit: {predicted_digit}")
    plt.axis("off")
    plt.show()

    return predicted_digit

image_paths = ["/content/1.jpg", "/content/2.jpg", "/content/3.png",
               "/content/4.jpg", "/content/5.jpg", "/content/6.png",
               "/content/7.jpg", "/content/8.jpg", "/content/9.png",
               "/content/10.jpg"]
for image_path in image_paths:
    predicted_digit = predict_digit(image_path)
    print(f'Số tiên đoán cho ảnh {image_path}: {predicted_digit}')
```
